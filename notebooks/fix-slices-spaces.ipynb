```python
import pandas as pd
```

```python
import re

with open("data/processed/omat_slices.txt", "r", encoding="utf-8") as f:
    content = f.read()

pattern = r'([o\-\+]) +([o\-\+])'
while re.search(pattern, content):
    content = re.sub(pattern, r'\1\2', content)

with open("data/processed/omat.txt", "w", encoding="utf-8") as f:
    f.write(content)
```

```python
import re

def fix_slices_in_df(df, column_name, save_file):
    pattern = r'([o\-\+]) +([o\-\+])'
    def patt(x):
        return re.sub(r'(?<=[o\-\+])\s+(?=[o\-\+])', '', x)
    df[column_name] = df[column_name].apply(patt)
    df.to_csv(save_file)
```

```python
fix_slices_in_df(pd.read_csv("/root/Crystal-generator-using-SLICES/data/processed/MP20_train_with_slices.csv"), 
"slices_string", "data/processed/MP20-train-slices-trimmed.csv")
fix_slices_in_df(pd.read_csv("/root/Crystal-generator-using-SLICES/data/processed/MP20_val_with_slices.csv"), 
"slices_string", "data/processed/MP20-val-slices-trimmed.csv")
```
